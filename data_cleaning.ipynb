```python
import pandas as pd
```

```python
#1. Load
df = pd.read_csv("/car_prices.csv")
```

```python
print(df.isnull().sum())
```

```python
#2. Remove Duplicate
df.drop_duplicates(inplace=True)
```

```python
#3. Handle Missing Value
df.fillna("Unknown",inplace=True)
```

```python
print(df.isnull().sum())
```

```python
#4. Remove Spaces and make lower
df.columns = df.columns.str.strip().str.lower()
```

```python
df.head(5)
```

```python
#5. Standardize text in columns
df['transmission'] = df['transmission'].str.strip().str.title()
df['color'] = df['color'].str.strip().str.title()
df['interior'] = df['interior'].str.strip().str.title()
df['seller'] = df['seller'].str.strip().str.title()
df.head(3)
```

```python
df.shape
```

```python
#6. Extract and Clean Date from salesdate
df['saledate'] = pd.to_datetime(df['saledate'], errors='coerce')
```

```python
df['saledate'] = df['saledate'].dt.strftime('%d-%m-%Y')
print(df['saledate'].head(3))
```

```python
#7. Save
df.to_csv("cleaned_car_prices.csv", index=False)
print("Done ! Final Shape :",df.shape)
```

```python
## Summary of Changes

Dataset: Car Prices (car_prices.csv)
Original Shape: (558837, 16)
Final Shape: (99812, 16)

Changes Made:
1. Load
2. Removed duplicate rows
3. Filled missing values with "Unknown"
4. Renamed all columns to lowercase
5. Standardized text columns (make, model, color, interior, seller, transmission) to Title Case
6. Cleaned saledate column to dd-mm-yyyy format

Tools Used: Python, Pandas (Google Colab)
```
